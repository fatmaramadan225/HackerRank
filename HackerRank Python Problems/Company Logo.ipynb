```python
s = input()
txt = list(s)
se = set(s)
dic = {}
count = 0

for i in se:
    for x in txt:
        if i == x:
            count += 1
    dic[i] = count
    count = 0


sorted_list = sorted(dic.items(), key=lambda item: (-item[1], item[0]))

for a, b in sorted_list[:3]:
    print(a, b)
```
