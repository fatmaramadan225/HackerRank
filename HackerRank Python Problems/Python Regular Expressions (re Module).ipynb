You are given a string .
Your task is to find out whether  is a valid regex or not.

Input Format

The first line contains integer , the number of test cases.
The next  lines contains the string .

Constraints


Output Format

Print "True" or "False" for each test case without quotes.

Sample Input

2
.*\+
.*+
Sample Output

True
False
Explanation

.*\+ : Valid regex.
.*+: Has the error multiple repeat. Hence, it is invalid.

````python
import re

t = int(input())

for _ in range(t):
    pattern = input()
    try:
        re.compile(pattern)
        
        if re.search(r'(\*\+|\+\+|\?\+|\*\*|\+\*|\?\*)', pattern):
            print("False")
        else:
            print("True")
            
    except re.error:
        print("False")
````

# 🧠 Python `re` Module (Regular Expressions)

## 🔹 What is `re`?
The `re` module in Python stands for **Regular Expressions**.  
It is used for **searching, matching, extracting, and replacing text** using patterns.

---

## 🔹 What is Regex?
A **Regular Expression (Regex)** is a pattern used to describe a set of strings.

Examples:
- `"cat"` → matches the word *cat*
- `"\d+"` → matches one or more digits
- `"[a-z]"` → matches any lowercase letter

---

## 🔹 Common Functions in `re`

### 1️⃣ `re.search()`
Searches for the **first match anywhere** in the string.

```python
import re

text = "My number is 123"
result = re.search(r"\d+", text)

print(result.group())  # 123

2️⃣ re.match()

Checks for a match only at the beginning of the string.

````python
re.match(r"\d+", "123abc")  # Match
re.match(r"\d+", "abc123")  # No match
````

3️⃣ re.findall()

Returns all matches as a list.

````python
re.findall(r"\d+", "a1 b22 c333")
# Output: ['1', '22', '333']
````

4️⃣ re.sub()

Replaces matches with another value.

````python
re.sub(r"\d+", "#", "a1 b22 c333")
# Output: 'a# b# c#'
````

5️⃣ re.split()

Splits a string based on a pattern.

````python
re.split(r"\s+", "Hello   world")
# Output: ['Hello', 'world']
````

6️⃣ re.compile()

Compiles a regex pattern for reuse (better performance).

````python
pattern = re.compile(r"\d+")

pattern.search("abc123")
pattern.findall("1 22 333")
````

| Symbol | Meaning         |
| ------ | --------------- |
| `.`    | Any character   |
| `\d`   | Digit           |
| `\w`   | Letter or digit |
| `\s`   | Whitespace      |
| `^`    | Start of string |
| `$`    | End of string   |
| `+`    | One or more     |
| `*`    | Zero or more    |
| `?`    | Optional        |
